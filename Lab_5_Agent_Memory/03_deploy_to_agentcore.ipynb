# Deploy the GraphRAG Agent with Memory to AgentCore Runtime

> **Hosted Workshop Note:** During hosted OneBlink workshops, SageMaker roles do not have the IAM permissions required to run this notebook. You can review the code and concepts here, but deployment will only work in your own AWS account with the appropriate permissions configured.

In `01_short_term_memory.ipynb` you wrapped the Lab 4 GraphRAG agent with a short-term memory layer and ran it locally. This notebook deploys that **same memory-backed agent** as a managed service on Amazon Bedrock AgentCore Runtime, then invokes it over REST. The conversation is remembered across separate invocations, keyed on the AgentCore session.

**Architecture:**
```
User -> AgentCore Runtime -> [get_context] -> Strands Agent (Claude)
             -> semantic_search / graph_enriched_search -> Neo4j
             -> [add_message] -> Neo4j (Conversation / Message)
```

The agent code and dependencies are pre-built in the `agentcore_deploy/` directory:
- `agent.py` — a `BedrockAgentCoreApp` handler that wraps the GraphRAG tools with the short-term memory layer (`get_context` before each turn, `add_message` after)
- `graphrag_agent.py` — the reusable retrievers and `@tool` functions (canonical copy in Lab 4's `lib/`, copied in at deploy time)
- `pyproject.toml` — Python dependencies, including `neo4j-graphrag[bedrock]` for the retrievers and `neo4j-agent-memory[bedrock]` for memory

You'll learn how to:
- Fill the deployment package with your Neo4j credentials from `CONFIG.txt`
- Deploy using `direct_code_deploy` (no Docker required)
- Invoke the deployed agent across several turns and watch memory carry context between them

This deploys under a distinct runtime name (`graphrag_memory_agent`), so it sits alongside the Lab 4 agent (`graphrag_strands_agent`) rather than replacing it.

**Prerequisites:** Complete `01_short_term_memory.ipynb` first. Ensure `../CONFIG.txt` is configured (including `NEO4J_DATABASE` for the memory layer) and the Lab 1 seed load populated the graph with chunks, embeddings, and the `chunkEmbeddings` vector index.

## 2. Fill agent.py from CONFIG.txt

The `agent.py` template ships with placeholder constants for the Neo4j connection. Read your credentials from `../CONFIG.txt` and write them into `agent.py`, so you never retype the values you entered back in Lab 1. The deployed agent carries these values (acceptable for the ephemeral workshop Aura instance).

Note the extra `NEO4J_DATABASE` key here versus the Lab 4 deploy: the memory SDK names the target database explicitly on every session, so the deployed agent needs it (defaults to `neo4j`).

This cell is idempotent: it rewrites the constant lines in place, so you can safely re-run it after editing `CONFIG.txt`.

> **Note:** The deployed `agent.py` will contain your Neo4j password in plain text. The repo copy stays as a template with placeholder tokens; this cell rewrites it in place with your real values. Do not commit the filled copy back. The generated `.bedrock_agentcore.yaml` is already git-ignored.

````python
import re
import shutil
from pathlib import Path

from dotenv import dotenv_values

config = dotenv_values("../CONFIG.txt")

DEFAULTS = {
    "NEO4J_DATABASE": "neo4j",
    "MODEL_ID": "us.anthropic.claude-sonnet-4-6",
    "REGION": "us-east-1",
}
KEYS = [
    "NEO4J_URI",
    "NEO4J_USERNAME",
    "NEO4J_PASSWORD",
    "NEO4J_DATABASE",
    "MODEL_ID",
    "REGION",
]

values = {}
for key in KEYS:
    val = config.get(key) or DEFAULTS.get(key)
    if not val:
        raise ValueError(f"{key} is missing from ../CONFIG.txt")
    values[key] = val

agent_path = Path("agentcore_deploy/agent.py")
source = agent_path.read_text()

# Replace each `KEY = ...` assignment line, whether it currently holds a
# placeholder token or a previously written value. This keeps the cell re-runnable.
for key, val in values.items():
    source = re.sub(
        rf"^{key} = .*$",
        f"{key} = {val!r}",
        source,
        count=1,
        flags=re.MULTILINE,
    )

agent_path.write_text(source)

# agent.py imports the reusable retrievers/tools from graphrag_agent.py. The
# canonical copy lives in Lab 4's lib/; copy it into this deploy directory so
# direct_code_deploy (which bundles only agentcore_deploy/) can import it. This
# copy is git-ignored.
graphrag_src = Path("../Lab_4_GraphRAG_Agent/lib/graphrag_agent.py")
shutil.copy(graphrag_src, Path("agentcore_deploy/graphrag_agent.py"))

print("agent.py updated from CONFIG.txt:")
print(f"  NEO4J_URI:      {values['NEO4J_URI']}")
print(f"  NEO4J_USERNAME: {values['NEO4J_USERNAME']}")
print(f"  NEO4J_PASSWORD: {'*' * 8} (hidden)")
print(f"  NEO4J_DATABASE: {values['NEO4J_DATABASE']}")
print(f"  MODEL_ID:       {values['MODEL_ID']}")
print(f"  REGION:         {values['REGION']}")
print(f"Copied {graphrag_src} -> agentcore_deploy/graphrag_agent.py")
````

## 4. Configure and Deploy

AgentCore needs a `.bedrock_agentcore.yaml` config that specifies the entrypoint, deployment type, and AWS settings. Generate it programmatically to avoid interactive CLI prompts. `execution_role_auto_create: True` lets the toolkit create the runtime execution role for you.

The toolkit packages your code locally and cross-compiles the dependencies for Linux ARM64 using prebuilt wheels, so no Docker is required. Both `neo4j-graphrag` and `neo4j-agent-memory` ship as pure-Python wheels on PyPI, which install cleanly under this cross-compilation.

The agent name is `graphrag_memory_agent`, distinct from the Lab 4 `graphrag_strands_agent`, so this deploys as its own runtime.

````python
import os

import boto3
import yaml
from dotenv import load_dotenv

load_dotenv("../CONFIG.txt")

REGION = os.getenv("REGION", "us-east-1")

sts = boto3.client("sts")
account_id = sts.get_caller_identity()["Account"]

agent_dir = os.path.abspath("agentcore_deploy")
entrypoint = os.path.join(agent_dir, "agent.py")

AGENT_NAME = "graphrag_memory_agent"

config = {
    "default_agent": AGENT_NAME,
    "agents": {
        AGENT_NAME: {
            "name": AGENT_NAME,
            "language": "python",
            "entrypoint": entrypoint,
            "deployment_type": "direct_code_deploy",
            "runtime_type": "PYTHON_3_13",
            "platform": "linux/arm64",
            "source_path": agent_dir,
            "aws": {
                "account": account_id,
                "region": REGION,
                "execution_role_auto_create": True,
                "ecr_auto_create": False,
                "s3_auto_create": True,
                "network_configuration": {
                    "network_mode": "PUBLIC",
                },
                "protocol_configuration": {
                    "server_protocol": "HTTP",
                },
                "observability": {
                    "enabled": True,
                },
            },
        }
    },
}

config_path = os.path.join(agent_dir, ".bedrock_agentcore.yaml")
with open(config_path, "w") as f:
    yaml.dump(config, f, default_flow_style=False)

print(f"Wrote {config_path}")
print(f"  Agent:      {AGENT_NAME}")
print(f"  Account:    {account_id}")
print(f"  Region:     {REGION}")
print(f"  Deploy:     direct_code_deploy")
````

### Invoke via boto3 — the memory demo

For programmatic access, use the `bedrock-agentcore` boto3 client. First read the deployed agent ARN from the generated config, then run four turns in **one session** so short-term memory links them.

````python
import yaml

with open("agentcore_deploy/.bedrock_agentcore.yaml") as f:
    deploy_config = yaml.safe_load(f)

default_agent = deploy_config["default_agent"]
agent_config = deploy_config["agents"][default_agent]
AGENT_ARN = agent_config["bedrock_agentcore"]["agent_arn"]
AGENT_REGION = agent_config["aws"]["region"]

print(f"Agent:  {default_agent}")
print(f"ARN:    {AGENT_ARN}")
print(f"Region: {AGENT_REGION}")
````

````python
import json
import uuid

from botocore.config import Config

agentcore_config = Config(read_timeout=300)

# One session id shared across all turns, so the deployed agent's short-term
# memory links them. (Lab 4 used a fresh uuid per call; here a stable session is
# exactly what makes the memory carry context between invocations.)
DEMO_SESSION_ID = str(uuid.uuid4())


def invoke_agent(prompt: str, session_id: str = DEMO_SESSION_ID):
    """Invoke the deployed agent via boto3, pinned to one memory session."""
    client = boto3.client(
        "bedrock-agentcore",
        region_name=AGENT_REGION,
        config=agentcore_config,
    )

    response = client.invoke_agent_runtime(
        agentRuntimeArn=AGENT_ARN,
        runtimeSessionId=session_id,
        payload=json.dumps({"prompt": prompt, "session_id": session_id}).encode(),
        qualifier="DEFAULT",
    )

    content = "".join(
        chunk.decode("utf-8") for chunk in response.get("response", [])
    )

    try:
        return json.loads(content)
    except (json.JSONDecodeError, ValueError):
        return content


# The headline demo: four turns in one session. Q2 never names Apple and Q4
# refers to the whole conversation, so both only work if memory carried the
# earlier turns across separate AgentCore invocations.
QUESTIONS = [
    "Tell me about Apple's risk factors.",
    "What about their competitors?",
    "How do those risks compare to NVIDIA's?",
    "Summarize what we've discussed so far.",
]

print(f"Session: {DEMO_SESSION_ID}\n")
for question in QUESTIONS:
    print(f"=== {question} ===")
    print(invoke_agent(question))
    print()
````

## 6. Inspect the Memory in Neo4j

The conversation the deployed agent had is now durable in the graph. Here are the `Message` nodes for the demo session, in order, sitting in the same database as the SEC 10-K `Company` and `Chunk` nodes. Run this after the boto3 cell above so `DEMO_SESSION_ID` is set.

````python
from dotenv import dotenv_values
from neo4j import GraphDatabase

cfg = dotenv_values("../CONFIG.txt")
database = cfg.get("NEO4J_DATABASE") or "neo4j"

driver = GraphDatabase.driver(
    cfg["NEO4J_URI"],
    auth=(cfg["NEO4J_USERNAME"], cfg["NEO4J_PASSWORD"]),
)
with driver.session(database=database) as session:
    rows = session.run(
        """
        MATCH (c:Conversation {session_id: $session_id})-[:HAS_MESSAGE]->(m:Message)
        RETURN m.role AS role, m.content AS content, m.timestamp AS timestamp
        ORDER BY m.timestamp
        """,
        session_id=DEMO_SESSION_ID,
    )
    for r in rows:
        print(f"[{r['role']}] {r['content'][:100]}")
driver.close()
````

## Try it in the AgentCore console playground

You can also drive the deployed runtime from the **Bedrock AgentCore console**: open your `graphrag_memory_agent` runtime and use its test/playground panel. Send these prompts **in order, keeping the same session** (the console keeps one session id per test panel, which is what links the turns in memory):

1. `{"prompt": "Tell me about Apple's risk factors."}` — entity-specific, triggers `graph_enriched_search`.
2. `{"prompt": "What about their competitors?"}` — "their" resolves to Apple only from memory.
3. `{"prompt": "How do those risks compare to NVIDIA's?"}` — memory continuity plus a new entity traversal.
4. `{"prompt": "Summarize what we've discussed so far."}` — pure memory recall, no new retrieval needed.

If you start a new session, memory resets and question 2 will no longer know who "their" refers to.
